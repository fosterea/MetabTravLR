```python
from metab_processing.metab_travlr_config import PROJECT_DATA_DIR, DATA_DIR as METAB_DATA_DIR

import csv

def get_csv_header(file_path):
    with open(file_path, mode='r', encoding='utf-8') as f:
        reader = csv.reader(f)
        header = next(reader) # Fetches only the first row
        return header

header = get_csv_header(f'{METAB_DATA_DIR}/MERSCOPE/vz-ffpe-showcase/HumanBreastCancerPatient1/cell_by_gene.csv')
```
