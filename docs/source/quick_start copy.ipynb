```python
# Human Tonsil snRNAseq from SlideTags
# https://www.nature.com/articles/s41586-023-06837-4

adata = sc.read_h5ad('/global/scratch/users/fosterangus/MetabTravLR/Data/Xenium/Primary_Dermal_Melanoma/adata.h5ad')
adata
```

```python
adata.obs['cell_type'] = adata.obs['leiden_scVI_res_0.5']
adata.layers['raw_count'] = adata.X
```

```python
from SpaceTravLR.spaceship import SpaceShip
```

```python
# Note: Make sure the outdir matches the one in the launch.py file

spacetravlr = SpaceShip(
    name='Melenoma', 
    outdir='/global/scratch/users/fosterangus/MetabTravLR/Data/SpaceTravLRMelenomaTest',
    genes=['CD4', 'CD3E', 'FOXP3', 'IL2RA', 'CTLA4', 'IL10', 'TGFB1', 'ENTPD1']
)
```

```python
spacetravlr.setup_(adata, overwrite=False, run_commot=False) # Can change overwrite to True
```

```python
assert spacetravlr.is_everything_ok()
```

```python
spacetravlr.focus_genes = ['CD4', 'CD3E', 'FOXP3', 'IL2RA', 'CTLA4', 'IL10', 'TGFB1', 'ENTPD1']
```

```python
## Run this multiple times to spawn parallel workers
spacetravlr.spawn_worker(
    account='fc_wagnerlabfca',
    partition='savio4_gpu',
    qos='a5k_gpu4_normal',
    gres='gpu:A5000:1',
    job_name='SpaceTravLR',
    cpus_per_task=4,
    lifespan=0.5,
    python_path='/global/home/users/fosterangus/.conda/envs/spacetravlr/bin/python'
) 
```

```python
spacetravlr.fit()
```

```python
spacetravlr.setup_perturbations(
    adata=adata, 
    use_float16=True
)
```

```python
spacetravlr.perturb(
    target='FOXO1',
    propagation=4,
    gene_expr=0,
)
```
